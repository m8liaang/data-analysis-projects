```python
import pandas as pd
import plotly.express as px
```

# Most Affected U.S. States Data

```python
df = pd.read_csv('https://raw.githubusercontent.com/CSSEGISandData/measles_data/refs/heads/main/Top_states_time_series.csv')
```

## Data Restructuring and Preprocessing

```python
df.tail()
```

```python
florida_df = df[['week_start', 'week_end', 'FL_cases']].copy()
florida_rolling = florida_df['FL_cases'].rolling(window=7).mean()

combined_fl = pd.concat([florida_df['week_start'], florida_rolling], axis=1)
```

```python
virginia_df = df[['week_start', 'week_end', 'VA_cases']].copy()
va_rolling = virginia_df['VA_cases'].rolling(window=7).mean()

combined_va = pd.concat([virginia_df['week_start'], va_rolling], axis=1)
```

```python
texas_df = df[['week_start', 'week_end', 'TX_cases']].copy()
tx_rolling = texas_df['TX_cases'].rolling(window=7).mean()

combined_tx = pd.concat([texas_df['week_start'], tx_rolling], axis=1)
```

```python
arizona_df = df[['week_start', 'week_end', 'AZ_cases']].copy()
az_rolling = arizona_df['AZ_cases'].rolling(window=7).mean()

combined_az = pd.concat([arizona_df['week_start'], az_rolling], axis=1)
```

```python
utah_df = df[['week_start', 'week_end', 'UT_cases']].copy()
ut_rolling = utah_df['UT_cases'].rolling(window=7).mean()

combined_ut = pd.concat([utah_df['week_start'], ut_rolling], axis=1)
```

```python
sc_df = df[['week_start', 'week_end', 'SC_cases']].copy()
sc_rolling = sc_df['SC_cases'].rolling(window=7).mean()

combined_sc = pd.concat([sc_df['week_start'], sc_rolling], axis=1)
```

```python
combined_df = pd.concat([df['week_start'], florida_rolling, va_rolling, tx_rolling, az_rolling, ut_rolling, sc_rolling], axis=1)

combined_df
```

## Data Visualization (Plotly)

```python
fig = px.bar(combined_fl, x=combined_fl['week_start'], y=combined_fl['FL_cases'],
             labels={'week_start' : 'Start Date of the Week',
                     'FL_cases' : 'Cases (Moving Average)'},
             title='Measles Cases in Florida')

fig.show()
```

```python
fig = px.bar(combined_va, x=combined_va['week_start'], y=combined_va['VA_cases'],
             labels={'week_start' : 'Start Date of the Week',
                     'VA_cases' : 'Cases (Moving Average)'},
             title='Measles Cases in Virginia')

fig.show()
```

```python
fig = px.bar(combined_ut, x=combined_ut['week_start'], y=combined_ut['UT_cases'],
             labels={'week_start' : 'Start Date of the Week',
                     'UT_cases' : 'Cases (Moving Average)'},
             title='Measles Cases in Utah')

fig.show()
```

```python
fig = px.bar(combined_sc, x=combined_sc['week_start'], y=combined_sc['SC_cases'],
             labels={'week_start' : 'Start Date of the Week',
                     'SC_cases' : 'Cases (Moving Average)'},
             title='Measles Cases in South Carolina')

fig.show()
```

```python
name_map = {
    'FL_cases' : 'Florida',
    'VA_cases' : 'Virginia',
    'TX_cases' : 'Texas',
    'AZ_cases' : 'Arizona',
    'UT_cases' : 'Utah',
    'SC_cases' : 'South Carolina',
}

fig = px.bar(combined_df, x=combined_df['week_start'], y=['FL_cases', 'VA_cases', 'TX_cases', 'AZ_cases', 'UT_cases', 'SC_cases'],
                labels={'variable' : 'State',
                        'week_start' : 'Start Date of Week',
                        'value' : 'Moving Average of Cases'},
                title='Measles Cases in Most Affected U.S. States')

for trace in fig.data:
  clean_name = name_map.get(trace.name, trace.name)
  trace.name = clean_name
  trace.hovertemplate = f"<b>{clean_name}</b><br>Start Date of Week=%{{x}}<br>Moving Average of Cases=%{{y}}<extra></extra>"

fig.show()
```

# All States Measles Data

```python
all = pd.read_csv('https://raw.githubusercontent.com/CSSEGISandData/measles_data/refs/heads/main/measles_county_all_updates_detailed.csv')
```

## Data Restructuring and Preprocessing

```python
all.info()
```

```python
all.head()
```

```python
all.tail()
```

```python
all.isna().any()
```

```python
all = all.drop(columns=['location_id', 'location_type'])

all
```

```python
states_names = ["Alabama", "Alaska", "Arizona", "Arkansas", "California", "Colorado", "Connecticut", "Delaware", "Florida", "Georgia", "Hawaii", "Idaho", "Illinois", "Indiana", "Iowa", "Kansas", "Kentucky", "Louisiana", "Maine", "Maryland", "Massachusetts", "Michigan", "Minnesota", "Mississippi", "Missouri", "Montana", "Nebraska", "Nevada", "New Hampshire", "New Jersey", "New Mexico", "New York", "North Carolina", "North Dakota", "Ohio", "Oklahoma", "Oregon", "Pennsylvania", "Rhode Island", "South Carolina", "South Dakota", "Tennessee", "Texas", "Utah", "Vermont", "Virginia", "Washington", "West Virginia", "Wisconsin", "Wyoming"]

filter = []

for state in states_names:
  state = all[all['location_name'].str.contains(state)]
  filter.append(state)

states = pd.concat(filter)
```

## Data Visualizations (Plotly)

```python
states['outcome_type'] = states['outcome_type'].replace({'case_imported' : 'Imported',
                                         'case_lab-confirmed' : 'Lab (Confirmed)',
                                         'case_local' : 'Local',
                                         'case_unvaccinated' : 'Unvaccinated',
                                         'case_vaccinated' : 'Vaccinated'})
```

```python
outcome = states.groupby('outcome_type', as_index=False)['value'].sum()

fig = px.bar(outcome, x='outcome_type', y='value',
             labels={'outcome_type' : 'Case Type',
                     'value' : 'Count'},
             title='Measles Case Type in U.S. (Imported, Local, Vaccinated, Unvaccinated, Lab (Confirmed))')

fig.show()
```

```python
states_df = states.copy()

states_df[['County', 'State']] = states_df['location_name'].str.split(', ', expand=True)
states_df = states_df.drop(columns=['County', 'location_name'])

states_df = states_df.groupby('State', as_index=False)['value'].sum()
```

```python
fig = px.bar(states_df, x='State', y='value',
             labels={'value' : 'Number of Cases'},
             title='Number of Measles Cases in Every U.S. State')

fig.show()
```

```python
dates_df = states.copy()
dates_df[['County', 'State']] = dates_df['location_name'].str.split(', ', expand=True)
dates_df = dates_df.drop(columns=['County', 'location_name'])

dates_df = dates_df.groupby('date', as_index=False)['value'].sum()
```

```python
fig = px.bar(dates_df, x='date', y='value',
             labels={'date' : 'Date',
                     'value' : 'Number of Measles Cases'},
             title='Number of Measles Cases (2025-2026)')

fig.show()
```
